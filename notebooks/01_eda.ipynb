# 01 - Exploratory Data Analysis: Fake Bills

Dataset: Kaggle *fake-bills* (alexandrepetit881234). Each row is a banknote described by 6 geometric measurements,
labelled genuine (`True`) or fake (`False`).

```python
import sys, os
sys.path.append(os.path.abspath(".."))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.utils import load_data, TARGET_COLUMN

%matplotlib inline
sns.set_theme(style="whitegrid")

DATA_PATH = "../dataset/fake_bills.csv"
PLOTS_DIR = "../results/plots"
os.makedirs(PLOTS_DIR, exist_ok=True)

df = load_data(DATA_PATH)
df.head()
```

## 1. Shape, types and summary statistics

```python
print("Shape:", df.shape)
df.info()
```

```python
df.describe().T
```

## 2. Missing values

```python
missing = df.isna().sum()
print(missing[missing > 0] if missing.any() else "No missing values")
print("\nDuplicate rows:", df.duplicated().sum())
```

## 3. Class balance (1 = genuine, 0 = fake)

```python
counts = df[TARGET_COLUMN].value_counts().sort_index()
print(counts)
ax = counts.rename({0: "Fake", 1: "Genuine"}).plot(kind="bar", color=["tomato", "seagreen"], rot=0)
ax.set_title("Class distribution")
ax.set_ylabel("Number of bills")
plt.tight_layout()
plt.savefig(f"{PLOTS_DIR}/class_distribution.png", dpi=150)
plt.show()
```

## 4. Feature distributions by class

```python
features = [c for c in df.columns if c != TARGET_COLUMN]
plot_df = df.copy()
plot_df["Class"] = plot_df[TARGET_COLUMN].map({0: "Fake", 1: "Genuine"})
palette = {"Fake": "tomato", "Genuine": "seagreen"}

fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, col in zip(axes.ravel(), features):
    sns.histplot(data=plot_df, x=col, hue="Class", kde=True, ax=ax, palette=palette)
    ax.set_title(col)
plt.tight_layout()
plt.savefig(f"{PLOTS_DIR}/feature_distributions.png", dpi=150)
plt.show()
```

```python
fig, axes = plt.subplots(2, 3, figsize=(14, 7))
for ax, col in zip(axes.ravel(), features):
    sns.boxplot(data=plot_df, x="Class", y=col, hue="Class", ax=ax, palette=palette, legend=False)
    ax.set_title(col)
plt.tight_layout()
plt.savefig(f"{PLOTS_DIR}/feature_boxplots.png", dpi=150)
plt.show()
```

## 5. Correlation heatmap

```python
plt.figure(figsize=(8, 6))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation matrix")
plt.tight_layout()
plt.savefig(f"{PLOTS_DIR}/correlation_heatmap.png", dpi=150)
plt.show()
```
